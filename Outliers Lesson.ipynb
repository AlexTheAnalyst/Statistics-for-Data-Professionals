```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_style('whitegrid')
customers = pd.read_csv('/Volumes/AlexHD/2. Analyst Builder/1. Raw/Statistics Course/files/customers.csv')
customers
```

```python
ltv = customers['lifetime_value']
```

```python
# Box plot shows all of these visually
fig, ax = plt.subplots(figsize=(8, 4))
sns.boxplot(x=ltv, ax=ax, color='lightblue')
ax.set_xlabel('Lifetime Value')
ax.set_title('LTV Spread')
plt.tight_layout()
plt.show()
```

```python
# - Determine IQR

q1, q3 = customers['lifetime_value'].quantile([0.25,0.75])

q3
```

```python
iqr = q3 - q1

iqr
```

```python
upper_fence = q3 + (2 * iqr)

upper_fence
```

```python
lower_fence = q1 - (2 * iqr)

lower_fence
```

```python
outliers = customers[(customers['lifetime_value'] > upper_fence) | 
    (customers['lifetime_value'] < lower_fence)].sort_values('lifetime_value', ascending = False)
```
